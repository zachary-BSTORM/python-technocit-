#### Somme des diagonales

- La fonction reçoit une matrice carrée (autant de lignes que de colonnes) et renvoie la somme des
éléments de ses deux diagonales :
    - la diagonale principale, d'en haut à gauche à en bas à droite ;
    - la diagonale secondaire, d'en haut à droite à en bas à gauche.
    - Une case qui appartient aux deux diagonales n'est comptée qu'une seule fois

```python
def somme_diagonales(matrice: list[list[int]]) -> int:
    n = len(matrice) 

    total = 0

    for i in range(n): 
        total += matrice[i][i]

        total += matrice[i][n - 1 - i] 

    if n % 2 == 1:
        millieu = n // 2
        total -= matrice[millieu][millieu]
    return total


value = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
result = 25

print(somme_diagonales(value) == result)
```

- La fonction reçoit un entier strictement positif n et renvoie True si n
est un nombre heureux, False sinon.
- Pour savoir si un nombre est heureux, on répète l'opération suivante :
    - on remplace le nombre par la somme des carrés de ses chiffres (par
    - exemple 19 devient 1² + 9² = 82) ;
    - on recommence avec le nouveau nombre.
- Si on finit par tomber sur 1, le nombre est heureux. Sinon, les valeurs se
mettent à tourner en rond indéfiniment sans jamais passer par 1 : le
nombre n'est pas heureux.

```python
def est_heureux(n: int) -> bool:
    deja_vus = set()
       
    while n != 1:
        
        if n in deja_vus:
            return False

        deja_vus.add(n)
        
        somme = 0
        while n > 0 :
            chiffre = n % 10
            somme += chiffre * chiffre
            n = n // 10
        n = somme
    return True

print(est_heureux(19))
print(est_heureux(2))
```

La fonction reçoit une liste non vide d'entiers positifs ou
nuls. On démarre sur la première case (indice 0).
La valeur de chaque case indique la longueur maximale du
saut possible depuis cette case : depuis une case qui vaut
3, on peut avancer de 1, 2 ou 3 cases (ou ne pas bouger).
Depuis une case qui vaut 0, on ne peut plus avancer.
La fonction renvoie True s'il est possible d'atteindre la
dernière case, et False sinon.

```python
def peut_atteindre_fin(sauts: list[int]) -> bool:
    portee = 0
    
    for i in range(len(sauts)):
        
        if i > portee:
            return False
        
        portee = max(portee , i + sauts[i])
        
    return True

print(peut_atteindre_fin([0, 1]))
```

- Des maisons sont alignées dans une rue. La liste maisons donne la
somme d'argent disponible dans chacune, dans l'ordre de la rue (tous
les montants sont positifs ou nuls).
Un cambrioleur veut emporter le maximum d'argent, mais il ne peut
pas cambrioler deux maisons voisines, sinon l'alarme se déclenche.
La fonction renvoie le montant maximum qu'il peut emporter. Si la
liste est vide, elle renvoie 0.

```python
def cambrioleur_zac(maisons: list[int]) -> int:
    before_last = 0
    last = 0

    for amount in maisons:
        i_take = before_last + amount

        not_take = last

        best = max(i_take, not_take)

        before_last = last
        last = best

    return last

def cambrioleur(maisons: list[int]) -> int:
    if len(maisons) == 0:
        return 0
    
    if len(maisons) == 1:
        return maisons[0]
    if len(maisons) == 2:
        return max(maisons[0], maisons[1])
    
    return max(maisons[0] + cambrioleur(maisons[2:]), maisons[1] + cambrioleur(maisons[3:]))



print(cambrioleur_zac([1, 2, 3, 1]))  # 4
print(cambrioleur_zac([2, 8, 5, 2, 2, 80]))  # 90
#                  2 ,8, 7, 10,9, 80

print(cambrioleur([1, 2, 3, 1]))  # 4
print(cambrioleur([2, 8, 5, 2, 2, 80]))  # 90
```

```python
def rotation_90(matrice: list[list[int]]) -> list[list[int]]:
    n = len(matrice)

    for i in range(n):
        for j in range(i + 1, n):
            matrice[i][j], matrice[j][i] = matrice[j][i], matrice[i][j]

    for l in matrice:
        l.reverse()

    return matrice


def rotation_90_second(matrice):
    n = len(matrice)

    new = []

    for _ in range(n):
        new.append([0] * n)

    for i in range(n):
        for j in range(n):
            new[j][n - 1 - i] = matrice[i][j]
    return new


def rotation_90(m):
    return [list(r) for r in zip(*m[::-1])]


print(rotation_90([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
print(rotation_90_second([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))


l_1 = [1, 2, 3]
l_2 = [4, 5, 6]
l_3 = [7, 8, 9]

result = []

for r in zip(l_1, l_2, l_3):
    result.append(list(r)[::-1])

for l in result:
    print(l)
```
