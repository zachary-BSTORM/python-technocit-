#### Enum
- On importe Enum depuis le module standard enum , puis on crée une classe qui en hérite.
Chaque attribut de la classe devient un membre de l'énumération

- Chaque membre a un nom ( .name ) et une valeur ( .value ). La valeur peut être de
n'importe quel type : chaîne, nombre, tuple…

- Par convention, les noms des membres sont en MAJUSCULES, comme des constantes.


```python
from enum import Enum,auto

class EtatFacture(Enum):
    BROUILLON = auto()
    COMPTABILISEE = auto()
    ANNULEE = auto()
    
    @property
    def etat_enum(self):
        etats = {
                EtatFacture.BROUILLON : "modifiable",
                EtatFacture.COMPTABILISEE : "complete",
                EtatFacture.ANNULEE : "annulation"
        }
        return etats[self]
    
    def info_state(self):
        match self:
            case EtatFacture.BROUILLON:
                return "Facture modifiable"
            case EtatFacture.COMPTABILISEE:
                return "Facture déja cloturé"
            case EtatFacture.ANNULEE:
                return "Facture annulée veuillez contactez le client"
    
    

etat = EtatFacture.ANNULEE


    
print(etat)
print(etat.value) # modifiable
print(etat.name) # BROUILLON


print(etat == "modifiable")
print(etat == EtatFacture.BROUILLON)

print(etat.value == "modifiable")

print(etat.etat_enum)

print(etat.info_state())

```

#### auto

- l'utilisation de **auto()** permet de générer un compteur pour les valeurs de l'enum

```python
from enum import Enum,auto


class Priority(Enum):
    BASSE = auto()
    NORMALE = auto()
    HAUTE = auto()
    

state_low = Priority.BASSE
state_medium = Priority.NORMALE
state_high = Priority.HAUTE

print(state_low.value)
print(state_medium.value)
print(state_high.value)
```
