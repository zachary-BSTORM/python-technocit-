# Le typage en python

````python
value_int = 1
print(value_int)
print(type(value_int))

value_float = 3.14
print(value_float)
print(type(value_float))

value_bool = True
print(value_bool)
print(type(value_bool))

value_str = "message"
print(value_str)
print(type(value_str))

value_str_multi = """
un texte
sur plusieurs
lignes
"""
print(value_str_multi)
print(type(value_str_multi))

# concatenation chaine de caractères
nom = "bob"
concat_un = f"du texte et {nom}"
print(concat_un)

# texte brut
text_brut = r"C:\Users\new"
print(text_brut)

# None absence de valeur 
inconnu = None
````

# Conversion de type en python

````python
value : str = "152.54"
# int
int()
value_int = int(value)

# float
float()
value_float = float(value)

# bool
bool()
value_bool = bool(value)

# str
str()
value_str = str(value_float)
````
