# Le typage en python

````python
# Entier : int
value_int = 1
print(value_int)
print(type(value_int))


# Décimal : float
value_float = 3.14
print(value_float)
print(type(value_float))


# Booléen : bool
value_bool = True
print(value_bool)
print(type(value_bool))


# Chaine de caractère : str
value_str = "message"
print(value_str)
print(type(value_str))

value_str_multi = """
un texte
sur plusieurs
lignes
"""
print(value_str_multi)
print(type(value_str_multi))


# concatenation chaine de caractères
nom = "bob"
concat_un = f"du texte et {nom}"
print(concat_un)

# texte brut ( n'interprète pas les caractères spéciaux comme \n)
text_brut = r"C:\Users\new"
print(text_brut)

# None absence de valeur 
inconnu = None
````
