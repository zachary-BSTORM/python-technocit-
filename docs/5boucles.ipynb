# Les boucles

## While

```python
continue_loop = True

while continue_loop:
    print("le code s'exécute")
    print('Attention au boucles infinies')
    
    input_user = input("Voulez vous continuer ? y - n :")
    if input_user == "n":
        break
    
i = 0
while i < 10:
    print(i)
    i += 1
```

## For

```python
collection = [1,2,3,4,5]

for v in collection:
    print(v)
```

```python
# range(depart,fin,pas)
# fin non compris
for i in range(0,11,1):
    print(i)
```

## Enumerate

```python
ma_collection = ["a","b","c","d"]

for i,value in enumerate(ma_collection,1):
    print(f"l'element {i} : {value}")
```

# Break & Continue

- break : met fin à la boucle actuelle

- continue : passe à l'itération suivante
