#### Staticmethod

- Une methode statique ne reçoit ni self ni cls 

- elle sera surtout utilisé pour des methodes utilitaires 

- Le décorateur **@staticmethod** qui permet d'appeler la methode sans cls ou self



```python
class Convertisseur:
    TAUX_YUAN = 7.80
    TAUX_DOLLARS = 1.08
    TAUX_LIVRE = 0.85
    
    
    @staticmethod
    def euro_to_dollars(montant):
        return round(montant * Convertisseur.TAUX_DOLLARS,2)
    
    @staticmethod
    def euro_to_yuan(montant):
        return round(montant * Convertisseur.TAUX_YUAN,2)
    
    @staticmethod
    def euro_to_livre(montant):
        return round(montant * Convertisseur.TAUX_LIVRE,2)
    
    
value = 10 # 10 euros

result = Convertisseur.euro_to_dollars(value) # appeler depuis la classe

print(result)

c = Convertisseur()

# appelé depuis une instance grace à "@staticmethod" 
# qui permet de ne pas passer self en premier paramètres
result_instance = c.euro_to_dollars(value) 


Convertisseur.euro_to_dollars()
Convertisseur.euro_to_yuan()
Convertisseur.euro_to_livre()

# il sera possible de n'importer que ce dont on à besoin 
# from convertisseur import euro_to_dollars
```
