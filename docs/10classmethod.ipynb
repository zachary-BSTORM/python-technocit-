#### Class method

- Une methode classique reçoit **self** 

- une methode de classe reçoit **cls**

- la methode sera précédé du décorateur **@classmethod**

- La methode pourra être appelé depuis la classe : **Pizza.class_value**



```python
class Pizza:
    diametre_default = 33
    
    recettes = {
        "margherita" : ["tomate" , "mozzarella"]
    }
    
    def __init__(self,nom,ingredients):
        self.nom = nom
        self.ingredients = ingredients
        self.diametre = Pizza.diametre_default
        
    @classmethod
    def change_diametre(cls,value):
        if 20 < value < 50:
            cls.diametre_default = value
        else:
            raise ValueError("Le nouveau diamètre est invalide ( 20cm -> 50cm)")
        
    
    @classmethod
    def depuis_nom(cls , nom):
            return cls(nom,cls.recettes[nom] ) 
            # return Pizza(nom,cls.recettes[nom] )  # ici le type sera toujours Pizza
        
class PizzaVeggi(Pizza):
    pass


# La valeur est accessible depuis la classe
print(Pizza.diametre_default)

# ou depuis une instance
pizza_margherita = PizzaVeggi.depuis_nom("margherita")
print(pizza_margherita.diametre)



# Si on modifie la valeur depuis la classe pas de soucis 
# Depuis l'instance on crée un attribut qui masque celui de la classe
Pizza.diametre_default = 34
Pizza.change_diametre(40)

pizza_margherita.diametre_default = 38


print(Pizza.diametre_default)
print(pizza_margherita.diametre_default)


# en utilisant le cls les classe enfant auront bien le type réel 
# en utilisant le nom de la classe : Pizza(nom) le type sera toujours le même

print(type(pizza_margherita).__name__)
        

```

```python
class Pizza:
    # Attribut de classe "privé" : _Pizza__diametre_default
    __diametre_default = 33

    def __init__(self, nom, ingredients):
        self.nom = nom
        self.ingredients = ingredients
        self.diametre = Pizza.__diametre_default

    # Méthode de classe
    @classmethod
    def diametre_par_defaut(cls):
        return cls.__diametre_default

    # Property : s'appelle sur un objet
    @property
    def diametre_par_defaut_objet(self):
        return self.__diametre_default


base_pizza = Pizza("base simple", ["tomates", "mozzarella"])

print(Pizza.diametre_par_defaut())           # 33  → classmethod, sur la classe
print(base_pizza.diametre_par_defaut())      # 33  → classmethod, depuis un objet
print(base_pizza.diametre_par_defaut_objet)  # 33  → property, sur un objet, sans ()
print(base_pizza.diametre)                   # 33  → attribut d'instance rempli par __init__
```
