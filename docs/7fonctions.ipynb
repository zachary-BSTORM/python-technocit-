# Les sous-programmes

Une **fonction** permet d'encapsuler un bloc de code réutilisable.  
Au lieu de réécrire la même logique 10 fois, on l'écrit une fois dans une fonction et on l'appelle quand on en a besoin.

On déclare une fonction avec le mot-clé **`def`**.

| Type | Description | Mot-clé |
|---|---|---|
| **Fonction** | Renvoie une valeur | `return` |
| **Procédure** | Effectue une action, ne renvoie rien | Pas de `return` |

> En Python, toute fonction sans `return` retourne automatiquement `None` (= rien, vide).

```python
# déclaration de fonction
def afficher_message():
    print("hello")

# appel de fonction
afficher_message()


def return_message():
    return "hello python"

result = return_message()

print(result)
```

```python
# avec paramètres
list_values = [1,2,3,4,5,6]

# ici le paramètre est obligatoire
def moyenne(values : list[int])-> float:
    total = 0
    for v in values:
        total += v
        
    return total / len(values)

moyenne(list_values)


# plusieurs paramètres dans un ordre différent
def afficher_animaux(animal_1,animal_2):
    print(f"Le premier animal est {animal_1}")
    print(f"Le deuxième animal est {animal_2}")
    
afficher_animaux(animal_2="rex",animal_1="garfield")
    
```

```python
# Valeur par défaut

def presentation(nom,age,poste="employé"):
    print(f"la personne {nom} agé de {age} est {poste}")
    
presentation("john",35)
presentation("Jane",35,"CEO")

# valeur par défaut ! risque 
def ajout_mauvais(valeur,liste=[]):
    liste.append(valeur)
    return liste

ajout_mauvais(10)
ajout_mauvais(15)
result = ajout_mauvais(20)

print(result)


def ajout_correct(valeur,liste = None):
    if liste is None:
        liste = list()
    liste.append(valeur)
    
    return liste

ajout_correct(10)
ajout_correct(15)
result = ajout_correct(20)
print(result)
```

# *args
- nombre indéfinie de paramètres 

```python
def moyenne(*args):
    total = sum(args)
    print(args)
    return total / len(args)

moyenne(1,2,3,4,5,6)
moyenne(1,2,3,4,5,67,8,9,9,9,6,5,576)
```

# **kwargs

- nombre indéfinie de paramètres nommés

```python
def creer_personne(**kwargs):
    new_personne = dict()
    print(kwargs)
    for k,v in kwargs.items():
            new_personne[k] = v
        
    return new_personne

creer_personne(name="Jane",prenom="Doe",age=30)
creer_personne(sport="tennis",plat="spaghetti",boisson="coca")
```

# Ordre à respecter

```python

# Fonction avec tous les paramètres : simple , facultatifs , args et kwargs
def fonction_full(name : str,age:int,poste="employé",*args,**kwargs):
    print(f"name : {name} - age : {age} {poste}")
    
    for v in args:
        print(v)
        
    for k,v in kwargs.items():
        print(f"k : {k} - v : {v}")
        
fonction_full("John",30,"Ceo",1,2,3,4,5,informations="allergique au noix")

# Sans args mais avec un tuple si l'ordre est modifié tout les paramètres doivent être nommés
def fonction_full(name : str,age:int,tuple_value,poste="employé",**kwargs):
    print(f"name : {name} - age : {age} {poste}")
    
    for v in tuple_value:
        print(v)
        
    for k,v in kwargs.items():
        print(f"k : {k} - v : {v}")
        
fonction_full(age=30,poste="Ceo",nom="John",tuple_value=(1,2,3,4,5),informations="allergique au noix")
```

# Quick sort

```python
import timeit

def quick_sort(liste : list[int]) -> list[int]:
    """
    liste : liste non trié
    
    result : renvoie la liste trié avec le quick-sort
    """
    if len(liste) <= 1:
        return liste
    
    pivot = liste.pop()
    petit = []
    grand = []
    
    for v in liste:
        if v < pivot:
            petit.append(v)
        else:
            grand.append(v)
            
    return quick_sort(petit) + [pivot] + quick_sort(grand)


values = [1,23,5,8,76,4,5,4,34,534,5,4,654,35,4,354,56,468,7,5,1,3,13,4,87,3,4,35]

print(quick_sort(values))


temps = timeit.timeit(
    stmt=lambda :quick_sort(values),
    number=5000
)

print(f"Temps total (5000) : {temps:.4f}s")
print(f"temps moyen par appel : {temps / 5000 * 1000000:2f}µs")
```
