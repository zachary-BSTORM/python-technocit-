```python
def fibo_simple(n):
    if n < 2:
        return n
    return fibo_simple(n - 1) + fibo_simple(n - 2)


print(fibo_simple(40))
```

```python
values = {}
def fibo(n):
    global values
    if n < 2:
        return 1
    if not n in values.keys():
        values[n] = fibo(n - 1) + fibo(n - 2)
    return values[n]



print(fibo(100))
```

```python
from functools import cache

@cache
def fibo_simple(n):
    if n < 2:
        return n
    return fibo_simple(n - 1) + fibo_simple(n - 2)


print(fibo_simple(978))
```
