# Convention de nommage en python

- Les nom doivent commencer par une lettre ou un underscore
- les noms ne peuvent pas commencer par un chiffe
- ils peuvent contenir que des lettres , chiffres et underscore
- la convention est le snake_case
- Les variables sont sensibles à la casse : a != A

# Affectation en python

- l'affectation se fait avec le symbol : **=**

# Pseudo code
- l'affectation se fait avec le **<-**

```python
# déclaration et type

a = 5
# le typage est uniquement pour documenter le code 
# il n'empeche pas de modifier le type de la variable
b : int = "test"

print(a)
print(b)

print(type(a))
print(type(b))
```
