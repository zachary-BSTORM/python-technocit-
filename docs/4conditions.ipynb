# Conditions en **Python**

n Python, on n'utilise **pas d'accolades** pour délimiter les blocs de code, mais l'**indentation** (généralement 4 espaces). Un bloc commence après les **deux points** `:`.

Les structures conditionnelles permettent d'exécuter du code **en fonction d'une condition**.

### Les mots-clés

| Mot-clé | Description |
|---|---|
| `if` | Si la condition est **vraie**, le bloc est exécuté |
| `elif` | Si la condition précédente est fausse, on teste cette nouvelle condition |
| `else` | Si **aucune** des conditions précédentes n'est vraie |
| `match` | Évalue la valeur d'une variable (disponible depuis **Python 3.10**, équivalent du `switch` d'autres langages) |

> Les parenthèses autour de la condition sont **optionnelles** mais peuvent améliorer la lisibilité.

```python
a = 5
b = 10

# if seul
if a < b:
    print("la valeur a est inférieur")
    
    
# if et elif le else n'est pas obligatoire
if a < b:
    print("la valeur a est inférieur")
elif b < a:
    print("la valeur b est inférieur")
    
    
# if elif et else 
if a < b:
    print("la valeur a est inférieur")
elif b < a:
    print("la valeur b est inférieur")
else:
    print('autre0')
    
    
# on peut utiliser plusieurs elif 
if a < b:
    print("la valeur a est inférieur")
    print("la valeur a est inférieur")
    print("la valeur a est inférieur")
    print("la valeur a est inférieur")
elif b < a:
    print("la valeur b est inférieur")
elif b == a:
    print("la valeur b est inférieur")
else:
    print('autre0')
    
# if et else uniquement ( un seul else ) 
if a < b:
    print("la valeur a est inférieur")
else:
    print('autre0')
```

# Match ( switch en js )
