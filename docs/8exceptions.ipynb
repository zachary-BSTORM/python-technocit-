# Gestion des erreurs

```python
try:
    pass # code à risque
except:
    pass # gestion de l'exception
else:
    pass # exécuté si pas d'erreur
finally:
    pass # executé dans tous les cas
```

````python
try:
    value = int(input("Entrez une valeur"))
except:
    print("La valeur est incorrecte")
````

---
## 2. Les exceptions courantes

Il faut **toujours préciser le type d'exception** — un `except:` nu attrape tout,
même les erreurs système (`KeyboardInterrupt`, `SystemExit`), ce qui masque des bugs réels.

| Exception | Cause typique |
|---|---|
| `ValueError` | Valeur incorrecte — `int("abc")` |
| `TypeError` | Type incompatible — `"a" + 1` |
| `ZeroDivisionError` | Division par zéro |
| `IndexError` | Index hors limites — `liste[999]` |
| `KeyError` | Clé absente dans un dict — `d["inexistante"]` |
| `AttributeError` | Attribut inexistant — `None.upper()` |
| `FileNotFoundError` | Fichier introuvable |
| `NameError` | Variable non définie |
| `RecursionError` | Trop d'appels récursifs |

# Récupération d'informations sur l'erreur

````python
try:
    x = int(input("Entrez une valeur : "))
except Exception as e: # gère toutes les erreurs
    print(f"Une erreur est survenue : {e}")
    print(f"Erreur inconnue : {type(e).__name__}") # affiche le type réel de l'erreur
except ValueError as e: # jamais atteint
    print(f"La valeur est incorrecte : {e}")
````

# Déclencher une erreur 

- raise : permet de lancer une exceptions

````python

def test_value(value : Any):
    
    match value:
        case int():
            raise Exception("C'est un entier")
        case str():
            raise Exception("C'est une chaine de caractères")
    

    
    
try:
    test_value("55")
    # test_value(55)
except Exception as e:
    print(f"Une erreur est survenue : {e}")
````
