## Opérateurs mathématiques

Les opérateurs mathématiques permettent d'effectuer des calculs sur les variables numériques.

| Opérateur | Description | Exemple |
|---|---|---|
| `+` | Addition | `10 + 5` → `15` |
| `-` | Soustraction | `10 - 5` → `5` |
| `*` | Multiplication | `10 * 5` → `50` |
| `/` | Division | `10 / 3` → `3.333...` |
| `//` | Division entière | `10 // 3` → `3` |
| `%` | Modulo (reste) | `10 % 3` → `1` |
| `**` | Puissance | `2 ** 3` → `8` |

```python
a = 10

b = 5

print(a + b) # addition
print(a - b) # soustraction
print(a * b) # multiplication
print(a / b) # division
print(a // b) # division entière
print(a % b) # modulo
print(a ** b) # puissance
```

## Opérateurs de comparaison

Les opérateurs de comparaison renvoient un **booléen** (`True` ou `False`).

| Opérateur | Description |
|---|---|
| `==` | Égal à |
| `!=` | Différent de |
| `<` | Inférieur à |
| `>` | Supérieur à |
| `<=` | Inférieur ou égal à |
| `>=` | Supérieur ou égal à |
| `in` | Appartient à (pour les collections) |
| `not in` | N'appartient pas à |
| `is` | Est le même objet en mémoire |
| `is not` | N'est pas le même objet |


```python
a = 10
b = 5

print(a == b)   # False — égal à
print(a != b)   # True  — différent de
print(a < b)    # False — inférieur à
print(a > b)    # True  — supérieur à
print(a <= b)   # False — inférieur ou égal à
print(a >= b)   # True  — supérieur ou égal à

# in / not in fonctionnent sur les collections
ma_liste = [1, 5, 10, 15]
print(10 in ma_liste)       # True  — 10 est dans la liste
print(20 not in ma_liste)   # True  — 20 n'est pas dans la liste

# is / is not comparent l'identité (même objet en mémoire)
print(a is b)       # False
print(a is not b)   # True
```

## Opérateurs logiques

Les opérateurs logiques permettent de combiner plusieurs conditions.

| Opérateur | Description | Exemple |
|---|---|---|
| `and` | ET logique — les deux conditions doivent être vraies | `True and False` → `False` |
| `or` | OU logique — au moins une condition doit être vraie | `True or False` → `True` |
| `not` | NON logique — inverse la valeur | `not True` → `False` |

```python
value1 = 10
value2 = 12

# and : et
print(value1 > 5 and value2 < 15) # True

# or : ou
print(value1 > 5 or value2 < 10) # True

# not : non
print(not (value1 > 5 and value2 < 15)) # False 
```

## Opérateurs d'affectation composée

Ces opérateurs combinent une opération et une affectation en une seule instruction.

| Opérateur | Équivalent | Description |
|---|---|---|
| `+=` | `c = c + a` | Addition et affectation |
| `-=` | `c = c - a` | Soustraction et affectation |
| `*=` | `c = c * a` | Multiplication et affectation |
| `/=` | `c = c / a` | Division et affectation |
| `//=` | `c = c // a` | Division entière et affectation |
| `%=` | `c = c % a` | Modulo et affectation |
| `**=` | `c = c ** a` | Puissance et affectation |

```python
a = 10
b = 5
c = None

# = : affectation
c = a + b # affectation de la valeur de a + b à c

# += : addition et affectation
c += a

# -= : soustraction et affectation
c -= a

# *= : multiplication et affectation
c *= a

# /= : division et affectation
c /= a

# //= : division entière et affectation
c //= a

# %= : modulo et affectation
c %= a

# **= : puissance et affectation
c **= a
```
